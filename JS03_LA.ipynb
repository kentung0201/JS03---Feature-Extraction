```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.feature_selection import SelectKBest, f_classif
from sklearn.metrics import accuracy_score, classification_report
```

```python
y = df["diagnosis"]

X = df.drop(columns=["diagnosis", "id", "Unnamed: 32"], errors="ignore")
```

```python
print("Jumlah fitur:", X.shape[1])
print("Nama fitur:")
print(X.columns.tolist())
```

```python
le = LabelEncoder()

y = le.fit_transform(y)
```

```python
print(y[:10])
```

```python
print(le.classes_)
```

```python
X.head()
```

```python
X.info()
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    stratify=y,
    test_size=0.2,
    random_state=42
)
```

```python
pipe = Pipeline([
    ("scaler", StandardScaler()),
    ("selector", SelectKBest(score_func=f_classif, k=5)),
    ("clf", LogisticRegression(max_iter=1000))
])
```

```python
pipe.fit(X_train, y_train)
```

```python
pred = pipe.predict(X_test)
```

```python
print("Accuracy:", accuracy_score(y_test, pred))
```

```python
print(classification_report(y_test, pred))
```

```python
selector = pipe.named_steps["selector"]

mask = selector.get_support()

selected_features = X.columns[mask]

print("Fitur yang dipilih:")
print(selected_features.tolist())
```

```python
results = []

for k in range(1, X.shape[1] + 1):

    pipe = Pipeline([
        ("scaler", StandardScaler()),
        ("selector", SelectKBest(score_func=f_classif, k=k)),
        ("clf", LogisticRegression(max_iter=1000))
    ])

    pipe.fit(X_train, y_train)

    pred = pipe.predict(X_test)

    accuracy = accuracy_score(y_test, pred)

    results.append((k, accuracy))
```

```python
results_df = pd.DataFrame(results, columns=["Jumlah Fitur", "Accuracy"])

results_df
```

```python
best_result = results_df.loc[results_df["Accuracy"].idxmax()]

print("Jumlah fitur optimal:", int(best_result["Jumlah Fitur"]))
print("Accuracy terbaik:", best_result["Accuracy"])
```

```python
best_k = int(best_result["Jumlah Fitur"])

best_pipe = Pipeline([
    ("scaler", StandardScaler()),
    ("selector", SelectKBest(score_func=f_classif, k=best_k)),
    ("clf", LogisticRegression(max_iter=1000))
])

best_pipe.fit(X_train, y_train)
```

```python
selector = best_pipe.named_steps["selector"]

mask = selector.get_support()

selected_features = X.columns[mask]

print("Jumlah fitur optimal:", best_k)
print("Fitur yang digunakan:")
for feature in selected_features:
    print("-", feature)
```

```python
pred = best_pipe.predict(X_test)

print("=== SelectKBest + Logistic Regression ===")
print("Jumlah fitur:", best_k)
print("Accuracy:", accuracy_score(y_test, pred))
print()
print(classification_report(y_test, pred))
```
